```python
def PatternCount(Text, Pattern):
    count = 0
    for i in range(len(Text) - len(Pattern) + 1):
        if Text[i:i+len(Pattern)] == Pattern:
            count += 1
    return count

def FrequentWords(Text, k):
    FrequentPatterns = set()
    Count = []

    for i in range(len(Text) - k + 1):
        Pattern = Text[i:i+k]
        Count.append(PatternCount(Text, Pattern))

    maxCount = max(Count)

    for i in range(len(Text) - k + 1):
        if Count[i] == maxCount:
            FrequentPatterns.add(Text[i:i+k])

    return FrequentPatterns
```

Análisis de Complejidad de FrequentWords(Text, k)
Supongamos que:

n = len(Text)

k es pequeño comparado con n

Paso a paso:
Primer for loop (línea 3–5):

Se ejecuta n - k + 1 veces.

En cada iteración, llama a PatternCount, que escanea todo el texto para contar el número de veces que aparece el patrón (longitud k).

Cada llamada a PatternCount cuesta O(n) tiempo.

🔁 Entonces este bloque tiene complejidad O(n²).

Calcular max(Count) (línea 6):

Esto es simplemente un escaneo del arreglo Count, de longitud n - k + 1.

Costo: O(n)

Segundo for loop (línea 7–9):

También recorre n - k + 1 elementos, y por cada uno puede agregar un patrón al conjunto.

Insertar en un set es O(1) en promedio.

Costo: O(n)

Remover duplicados (línea 10):

Ya usamos un set, así que esto es redundante o instantáneo.

💡 Conclusión de Complejidad
Tiempo total:
O(n²)
(el cuello de botella es la repetida llamada a PatternCount, que escanea todo el texto en cada iteración)

Espacio:

Count tiene tamaño O(n)

FrequentPatterns puede tener hasta O(n) patrones en el peor caso
→ O(n) espacio

```python
#VERSION OPTIMIZADA

def FrequentWordsFast(Text, k):
    freq_map = {}
    max_count = 0

    # Contar todos los k-mers en un solo paso
    for i in range(len(Text) - k + 1):
        pattern = Text[i:i+k]
        freq_map[pattern] = freq_map.get(pattern, 0) + 1
        if freq_map[pattern] > max_count:
            max_count = freq_map[pattern]

    # Recoger todos los patrones con frecuencia máxima
    FrequentPatterns = {pattern for pattern, count in freq_map.items() if count == max_count}

    return FrequentPatterns, max_count
```

```python
def FrequencyTable(Text, k):
    freqMap = {}
    n = len(Text)
    for i in range(n - k + 1):
        Pattern = Text[i:i+k]
        if Pattern not in freqMap:
            freqMap[Pattern] = 1
        else:
            freqMap[Pattern] += 1
    return freqMap
```

```python
def FrequentWordsFromTable(Text, k):
    freqMap = FrequencyTable(Text, k)
    maxCount = max(freqMap.values())
    FrequentPatterns = {pattern for pattern, count in freqMap.items() if count == maxCount}
    return FrequentPatterns, maxCount
```
